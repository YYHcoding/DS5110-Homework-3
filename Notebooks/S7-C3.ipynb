```python
import sqlite3
import pandas as pd
```

```python
con= sqlite3.connect("../data/movielens.db")
```

```python
df = pd.read_sql_query("SELECT * FROM movies",con)
df.head()
```

```python
schema =pd.read_sql_query("PRAGMA table_info(movies)",con)
types = schema[["name","type"]]
print("SQL column types")
display(types)
```

```python
print("Pandas column yypes")
display(df.dtypes)
```

```python
compare = pd.DataFrame({
    "SQlite type": schema["type"].values,
    "Pandas type": df.dtypes.astype(str).values
}, index=df.columns)

display(compare)
```

```python
sqlite_missing = {}
for i in df.columns:
    query = f'SELECT COUNT(*) FROM movies WHERE "{i}" IS NULL'
    sqlite_missing[i] = con.execute(query).fetchone()[0]

print("Missing values Sqlite")
display(pd.Series(sqlite_missing, name = "SQlite null count"))
```

```python
pandas_missing = df.isnull().sum()
print("Missing values in pandas")
display(pandas_missing)
```

```python
#This is for Step 7, case D
users = pd.read_sql_query("SELECT * FROM users",con)
count = (~users["zip"].astype("string").str.match(r"^\d",na=False)).sum()
print("ZIP codes not starting with digit: ",count)
```
